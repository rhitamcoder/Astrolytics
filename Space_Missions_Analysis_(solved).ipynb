# Use a Choropleth Map to Show the Number of Launches by Country

* Create a choropleth map using [the plotly documentation](https://plotly.com/python/choropleth-maps/)
* Experiment with [plotly's available colours](https://plotly.com/python/builtin-colorscales/). I quite like the sequential colour `matter` on this map.
* You'll need to extract a `country` feature as well as change the country names that no longer exist.

Wrangle the Country Names

You'll need to use a 3 letter country code for each country. You might have to change some country names.

* Russia is the Russian Federation
* New Mexico should be USA
* Yellow Sea refers to China
* Shahrud Missile Test Site should be Iran
* Pacific Missile Range Facility should be USA
* Barents Sea should be Russian Federation
* Gran Canaria should be USA


You can use the iso3166 package to convert the country names to Alpha3 format.

```python
# Extract the country from the Location column
df_data['Country'] = df_data['Location'].str.split(',').str[-1].str.strip()

# Fix locations that are not actual country names
country_replacements = {
    'Russia': 'Russian Federation',
    'New Mexico': 'USA',
    'Yellow Sea': 'China',
    'Shahrud Missile Test Site': 'Iran',
    'Pacific Missile Range Facility': 'USA',
    'Barents Sea': 'Russian Federation',
    'Gran Canaria': 'USA'
}

df_data['Country'] = df_data['Country'].replace(country_replacements)

df_data['Country'].value_counts()
```

```python
# Convert country names to 3-letter Alpha-3 country codes
df_data['Country_Code'] = df_data['Country'].apply(
    lambda country: countries.get(country).alpha3 if country in countries else None
)

# Count launches by country
launches_by_country = df_data['Country_Code'].value_counts().reset_index()
launches_by_country.columns = ['Country_Code', 'Launches']

# Create the choropleth map
fig = px.choropleth(
    launches_by_country,
    locations='Country_Code',
    color='Launches',
    color_continuous_scale='matter',
    title='Number of Space Launches by Country',
    labels={'Launches': 'Number of Launches'}
)

fig.show()
```

# Cold War Space Race: USA vs USSR

The cold war lasted from the start of the dataset up until 1991.

```python
# Filter the data for the Cold War Period (up to 1991)
cold_war_data = df_data[
    df_data['Date'].apply(lambda x: x.year) <= 1991
].copy()

# Group the launches by year and organisation
cold_war_launches = (
    cold_war_data
    .groupby([
        cold_war_data['Date'].apply(lambda x: x.year),
        'Organisation'
    ])
    .size()
    .reset_index(name='Launches')
)

cold_war_launches.head()
```

```python
fig = px.line(
    cold_war_launches,
    x='Date',
    y='Launches',
    color='Organisation',
    markers=True,
    title='Cold War Space Race: USA vs USSR',
    labels={
        'Date': 'Year',
        'Launches': 'Number of Launches',
        'Organisation': 'Organisation'
    }
)

fig.show()
```

## Create a Plotly Pie Chart comparing the total number of launches of the USSR and the USA

Hint: Remember to include former Soviet Republics like Kazakhstan when analysing the total number of launches.

```python
# Extract the country from the launch location
country = df_data['Location'].str.split(',').str[-1].str.strip()

# Count USA launches
usa_launches = (country == 'USA').sum()

# Count USSR launches, including Kazakhstan
ussr_launches = country.isin([
    'Russia',
    'Kazakhstan'
]).sum()

cold_war_totals = pd.Series({
    'USSR': ussr_launches,
    'USA': usa_launches
})

cold_war_totals
```

```python
fig = px.pie(
    values=cold_war_totals.values,
    names=cold_war_totals.index,
    title='Total Space Launches: USSR vs USA'
)

fig.show()
```

## Create a Chart that Shows the Total Number of Launches Year-On-Year by the Two Superpowers

```python
# Count yearly launches for the USSR and USA
cold_war_data['Year'] = cold_war_data['Date'].apply(lambda x: x.year)

usa_yearly = (
    cold_war_data[cold_war_data['Location'].str.split(',').str[-1].str.strip() == 'USA']
    .groupby('Year')
    .size()
)

ussr_yearly = (
    cold_war_data[cold_war_data['Location'].str.split(',').str[-1].str.strip().isin(['Russia', 'Kazakhstan'])]
    .groupby('Year')
    .size()
)

yearly_superpower_launches = pd.DataFrame({
    'USA': usa_yearly,
    'USSR': ussr_yearly
}).fillna(0)

yearly_superpower_launches
```

```python
fig = px.line(
    yearly_superpower_launches,
    x=yearly_superpower_launches.index,
    y=['USA', 'USSR'],
    markers=True,
    title='Year-on-Year Space Launches: USA vs USSR',
    labels={
        'x': 'Year',
        'value': 'Number of Launches',
        'variable': 'Superpower'
    }
)

fig.show()
```

## Chart the Percentage of Failures over Time

Did failures go up or down over time? Did the countries get better at minimising risk and improving their chances of success over time?

```python
# Calculate total missions and failed missions for each year
yearly_total = (
    df_data['Date']
    .apply(lambda x: x.year)
    .value_counts()
    .sort_index()
)

yearly_failures = (
    df_data[df_data['Mission_Status'] == 'Failure']['Date']
    .apply(lambda x: x.year)
    .value_counts()
    .sort_index()
)

yearly_failure_percentage = (
    yearly_failures / yearly_total * 100
).fillna(0)

yearly_failure_percentage
```

```python
fig = px.line(
    x=yearly_failure_percentage.index,
    y=yearly_failure_percentage.values,
    markers=True,
    title='Percentage of Mission Failures Over Time',
    labels={
        'x': 'Year',
        'y': 'Failure Percentage (%)'
    }
)

fig.show()
```

```python
fig = px.line(
    x=yearly_failure_percentage.index,
    y=yearly_failure_percentage.values,
    markers=True,
    title='Percentage of Mission Failures Over Time',
    labels={
        'x': 'Year',
        'y': 'Failure Percentage (%)'
    }
)

fig.update_yaxes(ticksuffix='%')
fig.show()
```

# For Every Year Show which Country was in the Lead in terms of Total Number of Launches up to and including including 2020)

Do the results change if we only look at the number of successful launches?

```python
# Find the country with the most launches for each year up to 2020
yearly_country_launches = (
    df_data[df_data['Date'].apply(lambda x: x.year) <= 2020]
    .assign(
        Year=lambda x: x['Date'].apply(lambda d: d.year),
        Country=lambda x: x['Location'].str.split(',').str[-1].str.strip()
    )
    .groupby(['Year', 'Country'])
    .size()
    .reset_index(name='Launches')
)

country_leader_each_year = (
    yearly_country_launches
    .sort_values(['Year', 'Launches'], ascending=[True, False])
    .drop_duplicates('Year')
    .sort_values('Year')
)

country_leader_each_year
```

```python
# Find the country with the most successful launches for each year up to 2020
successful_launches = df_data[
    (df_data['Mission_Status'] == 'Success') &
    (df_data['Date'].apply(lambda x: x.year) <= 2020)
].copy()

successful_country_leader = (
    successful_launches
    .assign(
        Year=lambda x: x['Date'].apply(lambda d: d.year),
        Country=lambda x: x['Location'].str.split(',').str[-1].str.strip()
    )
    .groupby(['Year', 'Country'])
    .size()
    .reset_index(name='Successful Launches')
    .sort_values(['Year', 'Successful Launches'], ascending=[True, False])
    .drop_duplicates('Year')
    .sort_values('Year')
)

successful_country_leader
```

# Create a Year-on-Year Chart Showing the Organisation Doing the Most Number of Launches

Which organisation was dominant in the 1970s and 1980s? Which organisation was dominant in 2018, 2019 and 2020?

```python
# Find the organisation with the most launches for each year
yearly_org_launches = (
    df_data[df_data['Date'].apply(lambda x: x.year) <= 2020]
    .assign(Year=lambda x: x['Date'].apply(lambda d: d.year))
    .groupby(['Year', 'Organisation'])
    .size()
    .reset_index(name='Launches')
)

organisation_leader_each_year = (
    yearly_org_launches
    .sort_values(['Year', 'Launches'], ascending=[True, False])
    .drop_duplicates('Year')
    .sort_values('Year')
)

organisation_leader_each_year
```

```python
# Plot the organisation doing the most launches each year
fig = px.bar(
    organisation_leader_each_year,
    x='Year',
    y='Launches',
    color='Organisation',
    title='Year-on-Year Organisation with the Most Space Launches',
    labels={
        'Year': 'Year',
        'Launches': 'Number of Launches',
        'Organisation': 'Organisation'
    }
)

fig.show()
```
